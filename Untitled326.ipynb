# PCA

```python
Principal componet analysis:

Feature reduction technique:

100  --->2,3 columns


it is an unsupervised ml technique used for dimensionality reduction 
it transforms original numeric features into new features
called principle components while preserving as much variance as possible


Student      Math Marks         Pysics    Chem       Math Marks+Pysics+Chem  =

A              80                 82         78
B              90                  89         99
C              40                42           38


How PCA works:

1.Collect numeric data
2.Standardize the feature
3.Calculate principle directions(it finds direction that capture the gretest variance
using eigen vectors and eigen values)

4.Select components
5.Transform the data

PC1=The direction that captures the maximum variance
PC2: The direction that capture maximum remaining variance while being perpendicular to pc1
PC3: the next direction perpendicular to the precedding components


A=[3  0
  0   2]

v1=[1
   0]

Av1=[3
    0]

Av1=3v1
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
import pandas as pd

X=pd.DataFrame(iris.data,
            columns=iris.feature_names)
```

```python
X
```

```python
y=iris.target
```

```python
y
```

```python
from  sklearn.preprocessing import StandardScaler
sc=StandardScaler()
```

```python
X_scaled=sc.fit_transform(X)
```

```python
X_scaled
```

# Apply PCA

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
X_reduced=pca.fit_transform(X_scaled)
```

```python
X_reduced
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(20,5))
for i ,name in enumerate(iris.target_names):
    plt.scatter(
        X_reduced[y==i,0],
        X_reduced[y==i,1],
        label=name
    )

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.grid()
```

```python
df=pd.DataFrame(X_reduced,columns=['P1','P2'])
```

```python
df
```

```python
from sklearn.linear_model import LogisticRegression

X=df
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(X,y,test_size=0.2,random_state=21)
```

```python
lr=LogisticRegression()
model=lr.fit(xtrain,ytrain)
```

# Prediction

```python
y_pred=model.predict(xtest)
```

```python
y_pred
```

```python
ytest
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest,y_pred)
```
